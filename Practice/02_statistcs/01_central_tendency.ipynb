# variance:

```python
# create a dataset:

data = np.array([2,3,4,5,6])
mean = np.mean(data)
variance = np.var(data,ddof=1)
variance
```

# standard deivation:

```python
df = np.array([600,470,170,430,300])
# calculate standard deviation
std_dev = np.std(df,ddof=1)
print("standard deviation is:", std_dev)
# calculate standard error:
std_error = std_dev/np.sqrt(len(df))
print("standard error is:", std_error)
```
