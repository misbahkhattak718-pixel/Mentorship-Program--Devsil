# Data distribution:

```python
# import libraries:
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
```

```python
# load the dataset:
df  = sns.load_dataset('tips')
df.head()
```

```python
# draw a histogram curve on it:
plt.figure(figsize=(10,6))
sns.histplot(data=df,x='total_bill', kde=True)
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df,x='tip', kde=True)
```

```python
# titanic dataset is loaded:
df_01  = sns.load_dataset('titanic')
df_01.head()
```

```python
# display basic information about the dataset:
df_01.info()
# check for missing values:
df_01.isnull().sum()
# display statistical summary of numerical columns:
df_01.describe()
```

```python
# display a histogram a titanic dataset 'age' column:
plt.figure(figsize=(10,6))
sns.histplot(data=df_01,x='age', kde=True)
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df_01,x='fare', kde=True, color='orange', bins=50,hue='class')
```

```python
# display a box plot for 'age' column:
plt.figure(figsize=(10,6))
sns.boxplot(data=df_01,x='age', color='lightgreen',hue='who')
```

# skewness and kortosis:


```python
df_01.head()
```

```python
# find the skewness and kurtosis of 'age' column:
stats.kurtosis(df_01['fare'],fisher=False)
```

```python
sns.histplot(data=df_01,x='fare', kde=True, color='orange', bins=50,hue='class')
```

```python
# find the skewness of 'fare' column: 
stats.skew(df_01['fare'])
```

```python
# find the skewness is negative or positive which is see the distributionn data plot is left or right skewed.
sns.histplot(data=df_01,x='fare', kde=True, color='lightblue', bins=50,hue='class')
stats.skew(df_01['fare'])
```
