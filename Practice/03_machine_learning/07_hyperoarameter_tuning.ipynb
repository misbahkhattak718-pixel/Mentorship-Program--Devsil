# Hyperparameter Tuning:

```python
Hyperparameter tuning is a proces of finding the best set
of hyperparameters for a machine learning model.
```

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
```

```python
# load the dataset
from sklearn.datasets import load_iris

iris = load_iris()
X = iris.data
y = iris.target
  
```

```python
%%time 
from sklearn.model_selection import GridSearchCV

# Define the model:

model = RandomForestClassifier()



# create the parameter grid:

param_grid = {
    'n_estimators': [10, 50, 100, 200],
    'max_depth': [None, 5, 10, 20],
    'criterion': ['gini', 'entropy'],
    'bootstrap': [True, False],
}

# set up the grid:
grid = GridSearchCV(estimator=model, param_grid=param_grid,cv=5, n_jobs=-1)


# fit the model:
grid.fit(X, y)

# print the best parameters:
print(f'Best parameters: {grid.best_params_}')
```

```python
%%time
from sklearn.model_selection import RandomizedSearchCV
# Define the model:

model = RandomForestClassifier()



# create the parameter grid:

param_grid = {
    'n_estimators': [10, 50, 100, 200],
    'max_depth': [None, 5, 10, 20],
    'criterion': ['gini', 'entropy'],
    'bootstrap': [True, False],
}


# set up the grid:
grid = RandomizedSearchCV(estimator=model, scoring='accuracy',param_distributions=param_grid,cv=5, n_jobs=-1)


# fit the model:
grid.fit(X, y)

# print the best parameters:
print(f'Best parameters: {grid.best_params_}')
```
