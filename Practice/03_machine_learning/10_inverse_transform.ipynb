# Inverse Transform/ Inverse  Encoding/ Decoding categorical variables:

`Inverse encoding`, also known as inverse transform or decoding categorical variables, is a 
technique used in data preprocessing to convert encoded categorical variables back to their
 original form. This is often necessary after applying encoding techniques like one-hot encoding, 
 label encoding, or ordinal encoding to prepare the data for machine learning models.

` Types of Inverse Encoding:`
1. `Inverse One-Hot Encoding:` This technique is used to reverse the one-hot encoding process, where each category is represented as a binary vector. The inverse process involves identifying which category corresponds to the binary vector and converting it back to its original categorical form.

2.` Inverse Label Encoding:` This technique is used to reverse the label encoding process, where each category is assigned a unique integer. The inverse process involves mapping the integers back to their corresponding categories.

3. `Inverse Ordinal Encoding: `This technique is used to reverse the ordinal encoding process, where categories are assigned integer values based on their order. The inverse process involves mapping the integer values back to their original categories.

4. `Inverse Binary Encoding:` This technique is used to reverse the binary encoding process, where categories are represented as binary digits. The inverse process involves converting the binary digits back to their original categorical form.

```python
# import libraries:
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split 
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler, MinMaxScaler, Normalizer,OrdinalEncoder
```

```python
# load the 'titanic' dataset:
df = sns.load_dataset('titanic')
# display the first 5 rows of the dataset:
df.head()
```

```python
# dont show the warnings:
import warnings
warnings.filterwarnings('ignore')


# impute the missing values fro age, deck , embarked,and embark_town:
df['age'].fillna(df['age'].mean(), inplace=True)

df['embarked'].fillna(df['embarked'].mode()[0], inplace=True)
df['embark_town'].fillna(df['embark_town'].mode()[0], inplace=True)

# drop the 'deck' column:
df.drop('deck', axis=1, inplace=True)
```

```python
# encoding all categorical variables and object variables:
from sklearn.preprocessing import LabelEncoder
le_sex = LabelEncoder()
le_class = LabelEncoder()
le_embarked = LabelEncoder()
le_who = LabelEncoder()
le_adult_male = LabelEncoder()
le_embark_town = LabelEncoder()
le_alive = LabelEncoder()
le_alone = LabelEncoder()

df['sex'] = le_sex.fit_transform(df['sex'])
df['class'] = le_class.fit_transform(df['class'])
df['embarked'] = le_embarked.fit_transform(df['embarked'])
df['who'] = le_who.fit_transform(df['who'])
df['adult_male'] = le_adult_male.fit_transform(df['adult_male'])
df['embark_town'] = le_embark_town.fit_transform(df['embark_town'])
df['alive'] = le_alive.fit_transform(df['alive'])
df['alone'] = le_alone.fit_transform(df['alone'])
df.head()
```

```python
# inverse transform  the data:
df['sex'] = le_sex.inverse_transform(df['sex'])
df['class'] = le_class.inverse_transform(df['class'])
df['embarked'] = le_embarked.inverse_transform(df['embarked'])
df['who'] = le_who.inverse_transform(df['who'])
df['adult_male'] = le_adult_male.inverse_transform(df['adult_male'])
df['embark_town'] = le_embark_town.inverse_transform(df['embark_town'])
df['alive'] = le_alive.inverse_transform(df['alive'])
df['alone'] = le_alone.inverse_transform(df['alone'])

df.head()
```
