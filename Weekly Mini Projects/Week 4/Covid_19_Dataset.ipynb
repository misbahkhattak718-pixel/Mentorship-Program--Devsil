```python
# preprocessing the each columns and new columns is add:
# convert the 'Country/Region' to 'Country':
df['Country'] = df['Country/Region']
df.drop('Country/Region',axis=1,inplace=True)
df.head()
```

```python
# the new columns is add:
df['New Deaths/100 Cases'] = df['New deaths'] / df['Confirmed']
df.head()
```

```python
# the new columns is add:
df['New Recovered/100 Cases'] = df['New recovered'] / df['Confirmed']
df.head()
```

```python
# the '1 week % increase' column is add:
df['1 week % increase'] = (df['Confirmed'] - df['Confirmed last week']) / df['Confirmed last week'] * 100
df.head()
```

```python
# the 'Confirmed last week'  is converted into 'Confirmed % last week':
df['Confirmed % last week'] = df['Confirmed last week'] / df['Confirmed'] * 100
df.head()
```
