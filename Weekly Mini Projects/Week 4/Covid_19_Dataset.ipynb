```python
# ANOVA: Used to compare means among more than two groups.

# ANOVA test used in  'Deaths' and 'Recovered' columns:
def anova_test (df, column1, column2):
    f_value, p_value = stats.f_oneway(df[column1], df[column2])
    return f_value, p_value
```

```python
# ANOVA test used in  'Deaths' and 'Recovered' columns:
f_value, p_value = anova_test(df, 'Deaths', 'Recovered')
print(f'F-value: {f_value}')
print(f'P-value: {p_value}')
```

```python
# print the results of ANOVA test used in''Deaths' and 'Recovered' columns use the if else statment:
if p_value < 0.05:
    print('The null hypothesis is rejected. There is a significant difference between the means of the two columns.')
else:
    print('The null hypothesis is not rejected. There is no significant difference between the means of the two columns.')
```


- why the null hypothesis is rejected? Because
- the p-value is 0.0000000000000002 which is less than 0.05.
- the p-value is less than 0.05, so the null hypothesis is rejected.
- there is a significant difference between the means of the two columns.
- the null hypothesis is rejected.

```python
# ANOVA test used in  'New Cases', 'New Deaths', 'New Recovered' columns:
f_value, p_value = anova_test(df, 'New Cases', 'New Deaths')
print(f'F-value: {f_value}')
print(f'P-value: {p_value}')

# print the results of ANOVA test used in''New Cases', 'New Deaths', 'New Recovered' columns use the if else statment:
if p_value < 0.05:
    print('The null hypothesis is rejected. There is a significant difference between the means of the two columns.')
else:
    print('The null hypothesis is not rejected. There is no significant difference between the means of the two columns.')
```

- why the null hypothesis is rejected?
- because the p-value is 0.0000000000000001 it is less than 0.05.
- there is a significant difference between the means of the two columns
- that is why the null hypothesis is rejected.
