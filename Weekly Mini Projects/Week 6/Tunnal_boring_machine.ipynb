## Project Description

### Introduction
This project revolves around analyzing the operational data from a tunneling machine used in construction. The data consists of various parameters recorded during drilling operations, such as stroke numbers, times, torque, and energy consumption. The goal is to assess the machine's efficiency, identify trends, and make recommendations for optimizing operations.

### Assumptions
In approaching this project, several assumptions were made:
- The data provided is accurate and represents typical operational conditions for the tunneling machine.
- External factors such as weather or geological changes are not considered, as the focus remains on the machine's performance.
- Sufficiently large datasets are available to draw meaningful conclusions.

## Methodology
The methodology applied in this project involves several steps:

1. **Data Collection**: Gather all relevant operational data from the tunneling machine, ensuring it's formatted correctly for analysis.
   
2. **Data Cleaning**: Identify and rectify any inconsistencies or errors in the data to ensure reliability in the findings.

3. **Exploratory Data Analysis (EDA)**:
   - **Statistical Analysis**: Calculate key metrics such as average stroke times, energy usage, and other performance indicators.
   - **Visualization**: Create graphs and charts to visually interpret trends and correlations within the data.
   - **Insights Extraction**: Identify patterns that could indicate inefficiencies or exceptional performance.
   
4. **Modeling**: Develop predictive models to estimate future performance based on historical data, utilizing techniques like regression analysis or machine learning algorithms.

5. **Recommendations**: Based on the analysis, offer actionable insights aimed at improving the efficiency of the tunneling operations.

## Data Description
The dataset comprises the following parameters:


| **Parameter**                   | **Description**                                           |
|----------------------------------|----------------------------------------------------------|
| **Stroke Number**                | An identifier for each drilling operation.               |
| **Date and Advance Time**        | The date of the operation and the duration it took.     |
| **Excavation per Stroke**        | The amount of material excavated with each drill stroke. |
| **Torque**                       | Measure of the rotational force exerted by the machine.  |
| **Thrust**                       | Measure of the linear force exerted by the machine.      |
| **Boring Energy**                | Energy consumed during the drilling process.             |
| **RPM (Revolutions per Minute)** | The speed of the drill.                                  |
| **Speed**                        | The operational speed of the machine.                    |
| **Gripper Pressure**             | The pressure applied by the grippers.                    |
| **Other Parameters**             | Include total revolutions, various pressures, and quality measures. |

This table provides an organized view of each parameter, making it easy to understand their meanings and significance in the context of the project.



## Objectives
The main objectives of the project are:

1. **Efficiency Assessment**: Evaluate the efficiency of the tunneling machine based on operational data to identify areas for improvement.

2. **Trend Analysis**: Understand trends over time to gauge whether performance improves, declines, or remains stable.

3. **Parameter Correlation**: Analyze how different operational parameters (e.g., torque, speed, and thrust) may affect overall performance.

4. **Actionable Insights**: Provide stakeholders with clear, actionable recommendations that can enhance the productivity and efficiency of future tunneling projects.

```python


# Display initial data types
print("Initial Data Types:")
print(df.dtypes)


# Define the expected data types
expected_types = {
    'Sr': 'int',
    'Stroke No.': 'int',
    'Advance Time': 'int',
    'Excavation per Stroke': 'float',
    'Boring Energy': 'float',
    'RPM': 'float',
    'Torque': 'float',
    'Thrust Force': 'float',
    'Speed': 'float',
    'Gripper Pressure': 'float',
    'Total Revolution': 'int',
    'Front Shield Pressure LHS': 'float',
    'Front Shield Pressure RHS': 'float',
    'Q-Value': 'float',
    'Q-Value based': 'float',
    'Penetration': 'float'
}

# Rename columns in the dataframe to match the expected column names
df.columns = [
    'Sr', 'Stroke No.', 'Date', 'Advance Time', 'Excavation per Stroke',
    'Boring Energy', 'RPM', 'Torque', 'Thrust Force', 'Speed',
    'Gripper Pressure', 'Total Revolution', 'Front Shield Pressure LHS',
    'Front Shield Pressure RHS', 'Q-Value', 'Q-Value based', 'Penetration'
]

# Convert columns to the expected data types and handle errors
for column, dtype in expected_types.items():
    try:
        df[column] = df[column].astype(dtype)
    except ValueError:
        print(f"Error converting {column}. Checking for invalid entries...")
        df[column] = pd.to_numeric(df[column], errors='coerce')

# Handle missing values
# Option 1: Fill missing values (e.g., with the mean or a specific value)
# For example, fill numeric columns with their mean
for column in expected_types:
    if df[column].dtype in ['float64', 'int64']:
        df[column].fillna(df[column].mean(), inplace=True)
    else:
        df[column].fillna(0, inplace=True)  # For integers or categorical data

# Option 2: Drop rows with missing values (Use cautiously)
# df.dropna(inplace=True)

# Identifying and removing duplicate rows
df.drop_duplicates(inplace=True)

# Display cleaned data types
print("\nCleaned Data Types:")
print(df.dtypes)

# Save the cleaned dataset to a new CSV file
df.to_csv('cleaned_dataset.csv', index=False)

print("\nData cleaning complete. Cleaned dataset saved as 'cleaned_dataset.csv'.")
```

```python


# Display initial data types
print("Initial Data Types:")
print(df.dtypes)

# Rename columns to match the expected column names (اگر پہلے سے ہیں تو نافذ نہیں ہوگا)
df.columns = [
    'Sr', 'Stroke No.', 'Date', 'Advance Time', 'Excavation per Stroke',
    'Boring Energy', 'RPM', 'Torque', 'Thrust Force', 'Speed',
    'Gripper Pressure', 'Total Revolution', 'Front Shield Pressure LHS',
    'Front Shield Pressure RHS', 'Q-Value', 'Q-Value based', 'Penetration'
]

# Ensure Date column is datetime
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Define the expected data types (Date included)
expected_types = {
    'Sr': 'int',
    'Stroke No.': 'int',
    'Date': 'datetime64[ns]',
    'Advance Time': 'int',
    'Excavation per Stroke': 'float',
    'Boring Energy': 'float',
    'RPM': 'float',
    'Torque': 'float',
    'Thrust Force': 'float',
    'Speed': 'float',
    'Gripper Pressure': 'float',
    'Total Revolution': 'int',
    'Front Shield Pressure LHS': 'float',
    'Front Shield Pressure RHS': 'float',
    'Q-Value': 'float',
    'Q-Value based': 'float',
    'Penetration': 'float'
}

# 1) First coerce every potential numeric column to numeric (to handle strings)
numeric_cols = list(expected_types.keys())
for col in numeric_cols:
    if expected_types[col] != 'datetime64[ns]':
        df[col] = pd.to_numeric(df[col], errors='coerce')

# 2) Now enforce the exact dtypes where applicable (except Date, which is already datetime)
for column, dtype in expected_types.items():
    if column == 'Date':
        # Already converted to datetime64[ns]
        continue
    try:
        df[column] = df[column].astype(dtype)
    except ValueError:
        # If direct cast fails, coerce again to the correct numeric type
        if dtype in ['float', 'int']:
            df[column] = pd.to_numeric(df[column], errors='coerce')
        else:
            df[column] = df[column].astype(dtype, errors='ignore')

# Handle missing values by imputing with mean for numeric columns
for column in expected_types:
    if df[column].dtype in ['float64', 'int64']:
        mean_value = df[column].mean()
        df[column].fillna(mean_value, inplace=True)

# Identifying and removing duplicate rows
df.drop_duplicates(inplace=True)

# Display cleaned data types
print("\nCleaned Data Types:")
print(df.dtypes)

# Save the cleaned dataset to a new CSV file
df.to_csv('cleaned_dataset.csv', index=False)

print("\nData cleaning complete. Cleaned dataset saved as 'cleaned_dataset.csv'.")
```

```python
# don't show the warnings
import warnings
warnings.filterwarnings('ignore')

# Check the outliers in the dataframe:
# Visualize the outliers in the dataframe using Boxplot:

plt.figure(figsize=(20, 20))

# Only include numeric columns for boxplot visualization
numeric_cols = df.select_dtypes(include=[np.number]).columns

re = 1
for i in numeric_cols:
    plt.subplot(len(numeric_cols) // 4 + 1, 4, re)
    sns.boxplot(x=df[i], palette="Set2")
    plt.title(i)
    re += 1

plt.tight_layout()
plt.show()
```

```python
# Remove the outlier in the dataset using Z-score method:
# Calculate Z-scores for numeric columns
z_scores = {}
for col in df.select_dtypes(include=[np.number]).columns:
    z_scores[col] = np.abs(stats.zscore(df[col]))

# Set a threshold for identifying outliers (commonly 3)
threshold = 3
# Create a boolean mask to identify outliers
outliers = np.zeros(df.shape[0], dtype=bool)
for col in z_scores:
    outliers |= (z_scores[col] > threshold)

# Remove outliers from the dataset
df_cleaned = df[~outliers]
print(f"Original dataset shape: {df.shape}")
print(f"Cleaned dataset shape: {df_cleaned.shape}")


```

```python
# visualize the outlier in the datafram :
# it confirm the outliers are removed through the boxplot visualization:
plt.figure(figsize=(20, 20))

# Only include numeric columns for boxplot visualization
numeric_cols = df_cleaned.select_dtypes(include=[np.number]).columns

re = 1
for i in numeric_cols:
    plt.subplot(len(numeric_cols) // 4 + 1, 4, re)
    sns.boxplot(x=df_cleaned[i], palette="Set2")
    plt.title(i)
    re += 1

plt.tight_layout()
plt.show()
```
