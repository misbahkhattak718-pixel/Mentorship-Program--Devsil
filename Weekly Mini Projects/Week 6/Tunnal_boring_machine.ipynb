## Project Description

### Introduction
This project revolves around analyzing the operational data from a tunneling machine used in construction. The data consists of various parameters recorded during drilling operations, such as stroke numbers, times, torque, and energy consumption. The goal is to assess the machine's efficiency, identify trends, and make recommendations for optimizing operations.

### Assumptions
In approaching this project, several assumptions were made:
- The data provided is accurate and represents typical operational conditions for the tunneling machine.
- External factors such as weather or geological changes are not considered, as the focus remains on the machine's performance.
- Sufficiently large datasets are available to draw meaningful conclusions.

## Methodology
The methodology applied in this project involves several steps:

1. **Data Collection**: Gather all relevant operational data from the tunneling machine, ensuring it's formatted correctly for analysis.
   
2. **Data Cleaning**: Identify and rectify any inconsistencies or errors in the data to ensure reliability in the findings.

3. **Exploratory Data Analysis (EDA)**:
   - **Statistical Analysis**: Calculate key metrics such as average stroke times, energy usage, and other performance indicators.
   - **Visualization**: Create graphs and charts to visually interpret trends and correlations within the data.
   - **Insights Extraction**: Identify patterns that could indicate inefficiencies or exceptional performance.
   
4. **Modeling**: Develop predictive models to estimate future performance based on historical data, utilizing techniques like regression analysis or machine learning algorithms.

5. **Recommendations**: Based on the analysis, offer actionable insights aimed at improving the efficiency of the tunneling operations.

## Data Description
The dataset comprises the following parameters:


| **Parameter**                   | **Description**                                           |
|----------------------------------|----------------------------------------------------------|
| **Stroke Number**                | An identifier for each drilling operation.               |
| **Date and Advance Time**        | The date of the operation and the duration it took.     |
| **Excavation per Stroke**        | The amount of material excavated with each drill stroke. |
| **Torque**                       | Measure of the rotational force exerted by the machine.  |
| **Thrust**                       | Measure of the linear force exerted by the machine.      |
| **Boring Energy**                | Energy consumed during the drilling process.             |
| **RPM (Revolutions per Minute)** | The speed of the drill.                                  |
| **Speed**                        | The operational speed of the machine.                    |
| **Gripper Pressure**             | The pressure applied by the grippers.                    |
| **Other Parameters**             | Include total revolutions, various pressures, and quality measures. |

This table provides an organized view of each parameter, making it easy to understand their meanings and significance in the context of the project.



## Objectives
The main objectives of the project are:

1. **Efficiency Assessment**: Evaluate the efficiency of the tunneling machine based on operational data to identify areas for improvement.

2. **Trend Analysis**: Understand trends over time to gauge whether performance improves, declines, or remains stable.

3. **Parameter Correlation**: Analyze how different operational parameters (e.g., torque, speed, and thrust) may affect overall performance.

4. **Actionable Insights**: Provide stakeholders with clear, actionable recommendations that can enhance the productivity and efficiency of future tunneling projects.

```python
# Check the missing values:
print(f"Missing values in each column:\n{df.isnull().sum().sort_values(ascending=False)}")
```

```python
# impute the missing values:
# For numerical columns, we can use the mean or median to fill missing values.
numerical_cols = df.select_dtypes(include=[np.number]).columns
for col in numerical_cols:
    df[col].fillna(df[col].mean(), inplace=True)
    
```

```python
# don't show the warnings

import warnings
warnings.filterwarnings("ignore")


# for categorical columns, we can use the mode to fill missing values.
categorical_cols = df.select_dtypes(include=["object"]).columns
for col in categorical_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```
