```python
# check the missing values again:
print(df.isnull().sum())
# handling missing values:
# For numerical columns, we can fill missing values with the mean or median.
numerical_columns = df.select_dtypes(include=[np.number]).columns
for column in numerical_columns:
    median_value = df[column].median()
    df[column].fillna(median_value, inplace=True)
# For categorical columns, we can fill missing values with the mode.
categorical_columns = df.select_dtypes(include=['object', 'category']).columns
for column in categorical_columns:
    mode_value = df[column].mode()[0]
    df[column].fillna(mode_value, inplace=True)
```

<div style="background-color:#E6F2FF; padding: 10px; border-radius: 8px; box-shadow: 0 1px 3px 0 rgba(0, 0, 0, 0.1); border:1px solid #66C2A5; margin-top: 10px; margin-left: 20px;">
    <h2 style="font-size:20px; font-family:Georgia, serif; color:#333333; text-align: left; margin: 0;">
        2.5 Handling Unreasonable Data Ranges
    </h2>
</div>

```python
# checking the handling umreasonable data ranges:
for column in numerical_columns:
    z_scores = np.abs(stats.zscore(df[column]))
    outliers = np.where(z_scores > 3)
    if len(outliers[0]) > 0:
        print(f'Outliers detected in column "{column}":')
        print(df.iloc[outliers])
        # Removing outliers:
        df = df[(z_scores < 3).all(axis=1)]
    else:
        print(f'No outliers detected in column "{column}".')
# Final check of the cleaned data:
print(df.isnull().sum())
print(df.describe(include='all'))
df.head()
```

```python
# preprocessing in each column and add new columns:
# Convert 'Yes'/'No' to boolean True/False in 'OverTime' column:
df['OverTime'] = df['OverTime'].map({'Yes': True, 'No': False})
df.head()
```

```python
# the 'HireDate' column to datetime format and extract 'HireYear' and 'HireMonth' and 'HireDay' columns:
df['HireDate'] = pd.to_datetime(df['HireDate'], errors='coerce')
df['HireYear'] = df['HireDate'].dt.year
df['HireMonth'] = df['HireDate'].dt.month
df['HireDay'] = df['HireDate'].dt.day
```

```python
print(df.head())
```

```python
# The 'YearsSinceLastPromotion' is conveted to specific ranges and create a new column 'PromotionCategory': 
def categorize_promotion(years):
    if years == 0:
        return 'Just Promoted'
    elif 1 <= years <= 3:
        return 'Recently Promoted'
    elif 4 <= years <= 7:
        return 'Moderately Promoted'
    else:
        return 'Not Promoted Recently'
df['PromotionCategory'] = df['YearsSinceLastPromotion'].apply(categorize_promotion)
```

```python
print(df.head())
```

```python
df.head()
```

```python
# The 'DistanceFromHome (KM)' is conveted to specific ranges and create a new column 'distanceCategory'
df['DistanceFromHome (KM)'].min(), df['DistanceFromHome (KM)'].max()
def categorize_distance(distance):
    if distance <= 5:
        return ''
    elif 6 <= distance <= 15:
        return 'Close'
    elif 16 <= distance <= 30:
        return 'Moderate'
    else:
        return 'Far'
df['DistanceCategory'] = df['DistanceFromHome (KM)'].apply(categorize_distance)
```

```python
# Remove the 'CommuteCategory' column as it is not necessary:
df.drop(columns=['CommuteCategory'], inplace=True)
```
