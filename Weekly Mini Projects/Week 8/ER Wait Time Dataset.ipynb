### 1. Introduction

***1.1 Project Background***

Emergency departments play a critical role in providing immediate medical care to patients with urgent health conditions. However, one of the most common challenges faced by hospitals is the long waiting time for patients before they receive medical attention. Long waiting times can lead to patient dissatisfaction, overcrowding, and reduced quality of healthcare services.

This project focuses on analyzing emergency room waiting time using hospital visit data. By studying factors such as urgency level, nurse-to-patient ratio, specialist availability, and facility size, the project aims to understand the reasons behind longer waiting times. Data analysis and machine learning techniques are used to identify patterns and predict waiting time so that hospitals can improve their services and reduce delays.

---

***1.2 Project Objectives***

The main objective of this project is to analyze and predict patient waiting time in emergency departments using data analysis and machine learning techniques.

The specific objectives of the project are:

To clean and prepare the dataset for analysis.

To explore the dataset and understand patient waiting time patterns.

To analyze how different factors such as urgency level, hospital size, and staffing affect waiting time.

To create meaningful visualizations that explain patient flow and waiting time distribution.

To build machine learning models that can predict total waiting time for patients.

To evaluate and compare the performance of different machine learning models.

To provide practical recommendations that can help hospitals reduce waiting time and improve patient satisfaction.

---

***1.3 Dataset Description***

The dataset used in this project contains information about emergency room visits from multiple hospitals. Each record in the dataset represents a single patient visit and includes details about the hospital, patient visit time, waiting stages, staffing conditions, and patient outcomes.

The dataset contains several important variables such as:

| No. | Variable Name                      | Description                                                                            |
| --- | ---------------------------------- | -------------------------------------------------------------------------------------- |
| 1   | Visit ID                           | Unique identifier for each hospital visit.                                             |
| 2   | Patient ID                         | Unique identifier for each patient.                                                    |
| 3   | Hospital ID                        | Unique identifier assigned to each hospital.                                           |
| 4   | Hospital Name                      | Name of the hospital where the patient visit occurred.                                 |
| 5   | Region                             | Indicates whether the hospital is located in an **Urban** or **Rural** area.           |
| 6   | Visit Date                         | Date and time when the patient arrived at the hospital.                                |
| 7   | Day of Week                        | Day when the hospital visit occurred (e.g., Monday, Tuesday).                          |
| 8   | Season                             | Season of the visit such as Winter, Spring, Summer, or Fall.                           |
| 9   | Time of Day                        | Indicates whether the visit occurred in the **Morning, Afternoon, Evening, or Night**. |
| 10  | Urgency Level                      | Severity level of the patient’s medical condition (Low, Medium, High).                 |
| 11  | Nurse-to-Patient Ratio             | Number of nurses available relative to the number of patients.                         |
| 12  | Specialist Availability            | Number of specialist doctors available during the visit.                               |
| 13  | Facility Size (Beds)               | Total number of beds available in the hospital.                                        |
| 14  | Time to Registration (min)         | Time taken for the patient to complete the registration process.                       |
| 15  | Time to Triage (min)               | Time taken for the patient to be assessed by triage staff.                             |
| 16  | Time to Medical Professional (min) | Time taken before the patient sees a doctor or medical professional.                   |
| 17  | Total Wait Time (min)              | Total waiting time from arrival until medical treatment begins.                        |
| 18  | Patient Outcome                    | Final result of the visit such as **Discharged or Admitted**.                          |
| 19  | Patient Satisfaction               | Rating given by the patient regarding their hospital experience.                       |
