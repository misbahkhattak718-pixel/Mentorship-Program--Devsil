### Project Introduction

This project focuses on analyzing social media engagement data to understand how users interact with different brands, products, and campaigns. The dataset contains information from platforms like Instagram, Twitter, Reddit, and YouTube.

The main goal is to use data analysis and machine learning techniques to find patterns in user behavior, sentiment, and engagement, and to predict engagement rates.

### Project Description

In this project, we:

Analyze how users react to different products and campaigns
Study engagement metrics like likes, shares, comments, and impressions
Explore how sentiment (positive/negative) affects engagement
Build machine learning models to predict engagement rate
Compare different models to find the best-performing one

This project helps businesses understand customer behavior and improve marketing strategies.

### Dataset Description

The dataset contains social media posts with multiple features:

### Basic Information


| **Category**           | **Feature Name**            | **Description**                                                       |
| ---------------------- | --------------------------- | --------------------------------------------------------------------- |
| **Basic Info**         | **post_id**                 | Unique identifier for each social media post                          |
|                        | **timestamp**               | Date and time when the post was created                               |
|                        | **day_of_week**             | Day of the week (Monday–Sunday)                                       |
|                        | **platform**                | Platform used (Instagram, Twitter, Reddit, YouTube)                   |
|                        | **user_id**                 | Unique ID of the user who posted                                      |
|                        | **location**                | Geographic location of the user                                       |
|                        | **language**                | Language in which the post is written                                 |
| **Content Features**   | **text_content**            | Main text/content of the post                                         |
|                        | **hashtags**                | Hashtags used to categorize or promote content                        |
|                        | **mentions**                | Tagged users or brands in the post                                    |
|                        | **keywords**                | Important keywords extracted from the text                            |
|                        | **topic_category**          | Category of content (e.g., Product, Pricing, Delivery, Support)       |
| **Sentiment Analysis** | **sentiment_score**         | Numeric value representing sentiment (positive or negative intensity) |
|                        | **sentiment_label**         | Overall sentiment (Positive / Negative)                               |
|                        | **emotion_type**            | Emotion expressed (Happy, Sad, Angry, Confused, Excited)              |
|                        | **toxicity_score**          | Measures harmful or offensive content level                           |
| **Engagement Metrics** | **likes_count**             | Number of likes received                                              |
|                        | **shares_count**            | Number of times the post was shared                                   |
|                        | **comments_count**          | Number of comments on the post                                        |
|                        | **impressions**             | Total number of views                                                 |
|                        | **engagement_rate**         | Overall engagement (interaction level relative to views)              |
| **Marketing Info**     | **brand_name**              | Brand associated with the post                                        |
|                        | **product_name**            | Product being discussed                                               |
|                        | **campaign_name**           | Name of the marketing campaign                                        |
|                        | **campaign_phase**          | Stage of campaign (Pre-Launch, Launch, Post-Launch)                   |
| **User Behavior**      | **user_past_sentiment_avg** | Average sentiment of user’s past posts                                |

```python
import matplotlib.pyplot as plt
# Set the style of the plot
plt.style.use('ggplot')
# Create a histogram of the 'Engagement' column
plt.figure(figsize=(10, 6))
plt.hist(df['engagement_rate'], bins=30, color='blue', edgecolor='black')
# Set the title and labels
plt.title('Distribution of Engagement', fontsize=16)
plt.xlabel('Engagement Rate', fontsize=14)
plt.ylabel('Frequency', fontsize=14)
# Show the plot
plt.show()
```

```python
# print the platform
print(df['platform'].unique())
```

```python
# Don not show the warnings
import warnings
warnings.filterwarnings("ignore")


# the platform  with the highest average engagement rate :
avg_engagement_by_platform = df.groupby('platform')['engagement_rate'].mean()
print(avg_engagement_by_platform)
```

```python
# Create a bar plot to visualize the average engagement rate by platform
plt.figure(figsize=(10, 6))
sns.barplot(x=avg_engagement_by_platform.index, y=avg_engagement_by_platform.values, palette='viridis')
# Set the title and labels
plt.title('Average Engagement Rate by Platform', fontsize=16)
plt.xlabel('Platform', fontsize=14)
plt.ylabel('Average Engagement Rate', fontsize=14)
# Show the plot
plt.show()
```

-  The upper plot is that the platform with the highest average engagement rate is Instagram, followed by Facebook and Twitter. This suggests that content on Instagram tends to receive more interactions (likes, comments, shares) compared to the other platforms in this dataset.
  
- The higher engagement rate on Instagram could be due to its visual nature, which often encourages more user interaction. Facebook and Twitter, while still popular, may not foster the same level of engagement for the content in this dataset.
  
- This insight can be valuable for marketers and content creators when deciding where to focus their efforts for maximum engagement.
